**Counting Words in a File**


Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
def count_text_file(file_path):

    with open(file_path, 'r') as file:
        lines = file.readlines()

        
        word_count = sum(len(line.split()) for line in lines)
    

        return  word_count


file_path = 'document.txt'

words = count_text_file(file_path)

print(f"Words: {words}")
```

**Reading a File in Reverse**


Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.


```python
def reverse_order():
    with open("reverse.txt","r") as file:
        lines= file.readlines()
    for line in lines[::-1]:
        print(line)

reverse_order()            
```

**Counting Lines, Words, and Characters**

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.


```python
def count_text_file(file_path):

    with open(file_path, 'r') as file:
        lines = file.readlines()

        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)

        return line_count, word_count, char_count


file_path = 'stats.txt'

lines, words, characters = count_text_file(file_path)

print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

**Data Aggregation** 

Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.


```python



data_frame = x.DataFrame({
    "Category": num.random.choice(["A", "B", "C"], 6),
    "Value": num.random.randint(1, 10, 6)
})

print(data_frame)

answer = data_frame.groupby("Category")["Value"].agg(["sum","mean"])

print(answer)
```

**Merging DataFrames**
